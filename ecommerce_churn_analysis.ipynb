แปลง Categorical เป็น numeric

```python
# สร้าง list ของชื่อคอลัมน์ที่เป็น categorical
cat_features = ['PreferredLoginDevice',	'PreferredPaymentMode',	'PreferedOrderCat']

# ลบ column ที่เดี๋ยวจะแปลงค่าภายหลังออกก่อน
Ecom_enc = df.drop(cat_features, axis=1) #ไม่ได้แก้ไข df ต้นฉบับ
Ecom_enc
```

```python
# เนื่องจาก Gender และ MaritalStatus มีแค่ 2 ค่า ไม่จำเป็นต้องใช้ OneHotEncoder สามารถใช้คำสั่ง map ได้
Ecom_enc['Gender'] = Ecom_enc['Gender'].map({'Male':0, 'Female':1})
Ecom_enc.head()
```

```python
Ecom_enc['MaritalStatus'] = Ecom_enc['MaritalStatus'].map({'Single':0, 'Married':1})
Ecom_enc
```

แปลง Category เป็น Numeric โดยใช้ OneHotEncoder จาก sklearn.preprocessing

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder()
tmp = encoder.fit_transform(df[cat_features]) # แปลงแต่ละค่าของ column ให้กลายเป็น one-hot vector ผลลัพธ์จะได้เป็น sparse matrix
tmp.toarray() # แปลง sparse matrix → dense array
```

```python
# รายการค่าที่เป็นไปได้ในแต่ละ feature ที่ OneHotEncoder เรียนรู้หลังจาก fit()
# ใช้ดูว่าแต่ละ feature มี category อะไรบ้าง
encoder.categories_
```

```python
# นำข้อมูลที่ transform แล้ว ไปใส่เพิ่มใน df
# สร้าง list ที่รวมชื่อ feature และชื่อ category ไว้ด้วยกัน
features = [col+'-'+cat
            for col, cats in zip(cat_features, encoder.categories_)
            for cat in cats]
features
```

```python
# เพิ่ม column ใน heart_enc
Ecom_enc[features] =tmp.toarray()
Ecom_enc
```

### Interpreting the Gradient Boosting Classifier Results

**Direct Class Predictions (`y_pred_gbc`):**
The `y_pred_gbc` array contains the predicted class labels for each instance in the `X_test` dataset. For a binary classification problem like customer churn, these predictions will be either 'Churn' or 'NotChurn' (as mapped from 1 and 0, respectively, in the original `y_train` and `y_test`). Each entry in `y_pred_gbc` represents the class that the `gbc` model believes is most likely for that particular customer based on its learned patterns.

For example, if `y_pred_gbc[0]` is 'NotChurn', it means the model predicts that the first customer in `X_test` will not churn.

**Class Probabilities (`y_pred_proba`):**
The `y_pred_proba` array provides a more nuanced understanding of the model's predictions. It's a 2D array where each row corresponds to a customer in `X_test`, and each column corresponds to the probability of that customer belonging to a specific class. In this case, since the `y_train` was mapped to `{'NotChurn': 0, 'Churn': 1}`, the columns in `y_pred_proba` will represent:
- The first column (`y_pred_proba[:, 0]`): Probability of the customer **not churning** (class 'NotChurn').
- The second column (`y_pred_proba[:, 1]`): Probability of the customer **churning** (class 'Churn').

The sum of probabilities for each row should be approximately 1.0. The model assigns a customer to a class based on which probability is higher (by default, if P(Churn) > P(NotChurn), it predicts 'Churn').

For example, if `y_pred_proba[0]` is `[0.02784452, 0.97215548]`:
- The probability of this customer being 'NotChurn' is approximately 2.78%.
- The probability of this customer being 'Churn' is approximately 97.22%.

This high probability for 'Churn' would lead the model to classify this customer as 'Churn' in the `y_pred_gbc` output.

These probabilities are crucial for understanding the model's confidence in its predictions and can be used to set custom thresholds for classification, which might be beneficial in scenarios where the costs of false positives and false negatives are different.

## การเปรียบเทียบระหว่าง Random Forest และ XGBoost

ทั้ง Random Forest และ Extreme Gradient Boosting (XGBoost) เป็นอัลกอริทึมประเภท Ensemble Learning ที่สร้างต้นไม้ตัดสินใจหลายต้นเพื่อทำการทำนาย แต่มีวิธีการรวมผลและจุดเด่นที่แตกต่างกัน

### 1. Random Forest

**แนวคิดหลัก:**
*   เป็นโมเดลประเภท **Bagging** (Bootstrap Aggregating) โดยการสร้างต้นไม้ตัดสินใจอิสระหลายต้น
*   แต่ละต้นไม้ถูกฝึกด้วยข้อมูลชุดย่อยแบบสุ่ม (Bootstrap Sample) จากชุดข้อมูลฝึกต้นฉบับ (Sampling with replacement)
*   ในการตัดสินใจแต่ละโหนดของต้นไม้ จะพิจารณาเฉพาะฟีเจอร์ย่อยแบบสุ่มเท่านั้น (Feature Randomness)
*   ผลลัพธ์สุดท้ายได้จากการรวม (averaging สำหรับ Regression หรือ majority voting สำหรับ Classification) ของผลลัพธ์จากทุกต้นไม้

**ข้อดี:**
*   **ลด Overfitting ได้ดี:** เนื่องจากสร้างต้นไม้หลายต้นและใช้การสุ่ม ทำให้โมเดลมีความเสถียรและ Generalize ได้ดี
*   **ใช้งานง่าย:** ไม่จำเป็นต้องปรับ Hyperparameters มากมายก็มักจะให้ผลลัพธ์ที่ดี
*   **จัดการ Missing Values ได้:** ไม่ค่อยอ่อนไหวต่อ Missing Values (หากจัดการอย่างเหมาะสม)
*   **ไม่อ่อนไหวต่อ Outliers:** เนื่องจากใช้ค่าเฉลี่ย/โหวตจากหลายต้นไม้
*   **ทำงานแบบ Parallel ได้ดี:** แต่ละต้นไม้สามารถสร้างและฝึกได้พร้อมกัน

**ข้อเสีย:**
*   **ประสิทธิภาพอาจไม่ถึงขีดสุด:** บางครั้งอาจไม่ได้ประสิทธิภาพสูงสุดเท่าโมเดล Boosting ถ้าปรับแต่งได้ไม่ดี
*   **ใช้ทรัพยากรมาก:** จำเป็นต้องใช้หน่วยความจำและเวลาในการฝึกสูงเมื่อจำนวนต้นไม้เพิ่มขึ้น
*   **ตีความยาก:** การตีความแต่ละต้นไม้เดี่ยวยาก เนื่องจากมีหลายต้น

### 2. Extreme Gradient Boosting (XGBoost)

**แนวคิดหลัก:**
*   เป็นโมเดลประเภท **Boosting** (โดยเฉพาะ Gradient Boosting) โดยสร้างต้นไม้ตัดสินใจแบบลำดับ (Sequentially)
*   ต้นไม้แต่ละต้นถูกสร้างขึ้นเพื่อแก้ไขข้อผิดพลาด (Residuals) ของต้นไม้ก่อนหน้า
*   มีการเพิ่มประสิทธิภาพด้วยเทคนิคหลายอย่าง เช่น Regularization (L1 & L2), Loss Function ที่ปรับแต่งได้, และเทคนิคการจัดการข้อมูลที่มีประสิทธิภาพ
*   มักจะใช้เป็นโมเดลในการแข่งขัน Machine Learning (เช่น Kaggle) เนื่องจากให้ประสิทธิภาพสูง

**ข้อดี:**
*   **ประสิทธิภาพสูง:** มักจะให้ผลลัพธ์ที่แม่นยำที่สุดในหลายๆ งาน
*   **ยืดหยุ่น:** รองรับ Objective Function และ Evaluation Metric ได้หลากหลาย
*   **จัดการ Regularization ได้ดี:** มีกลไกป้องกัน Overfitting ที่มีประสิทธิภาพ
*   **ทำงานได้เร็ว:** มีการปรับปรุงโค้ดและโครงสร้างให้มีประสิทธิภาพสูง สามารถใช้ CPU Cores ได้เต็มที่
*   **จัดการ Missing Values ได้:** มีกลไกจัดการ Missing Values ภายในตัว

**ข้อเสีย:**
*   **อ่อนไหวต่อ Overfitting:** หากปรับ Hyperparameters ไม่ดี อาจเกิด Overfitting ได้ง่ายกว่า Random Forest
*   **ใช้ทรัพยากรมาก:** การคำนวณซับซ้อนกว่าและใช้หน่วยความจำมาก โดยเฉพาะอย่างยิ่งเมื่อสร้างต้นไม้ลึกๆ
*   **ทำงานแบบ Parallel ได้จำกัด:** การสร้างต้นไม้แต่ละต้นต้องอาศัยผลจากต้นก่อนหน้า ทำให้การทำงานแบบ Parallel ในส่วนของต้นไม้ทำได้จำกัด (แต่มีการ Parallel ในส่วนอื่นๆ เช่น การหา Split Point)
*   **ตีความยาก:** เป็นโมเดลประเภท Black Box ที่ยากต่อการตีความผลลัพธ์โดยตรง

### สรุปความแตกต่างที่สำคัญ

| คุณสมบัติ          | Random Forest                                | XGBoost                                       |
| :---------------- | :------------------------------------------- | :-------------------------------------------- |
| **แนวคิด**        | Bagging (ต้นไม้สร้างอิสระ)                  | Boosting (ต้นไม้สร้างต่อเนื่องเพื่อแก้ไขข้อผิดพลาด) |
| **การลด Variance/Bias** | ลด Variance เป็นหลัก                        | ลด Bias เป็นหลัก                              |
| **ความเร็ว**      | เร็วในการฝึก (Parallel)                     | เร็วในการฝึก (Optimized), ช้ากว่า RF ในบางกรณี |
| **ประสิทธิภาพ**   | ดีมาก, มั่นคง                               | มักจะดีที่สุด, อาจ Overfit ได้ถ้าปรับไม่ดี     |
| **ความซับซ้อน**    | ต่ำถึงปานกลาง                                | สูงกว่า, มี Hyperparameters ให้ปรับเยอะ       |
| **Overfitting**   | ต้านทานได้ดี                                 | อ่อนไหวถ้าปรับไม่ดี                           |

โดยทั่วไป หากคุณต้องการโมเดลที่รวดเร็ว มั่นคง และจัดการได้ง่าย Random Forest เป็นตัวเลือกที่ดีเยี่ยม แต่ถ้าคุณต้องการประสิทธิภาพสูงสุดและพร้อมที่จะใช้เวลาในการปรับแต่ง Hyperparameters อย่างละเอียด XGBoost มักจะเป็นตัวเลือกที่ให้ผลลัพธ์ที่ดีที่สุดครับ

*   **How to interpret class probabilities predicted by the Gradient Boosting Classifier?**
    The class probabilities (`y_pred_proba`) offer a more detailed view, showing the likelihood of each customer belonging to each class. For a binary classification like customer churn, `y_pred_proba` is a 2D array where each row corresponds to a customer and the columns represent the probability of 'NotChurn' (first column) and 'Churn' (second column). The sum of probabilities for each customer row is approximately 1.0. The class with the highest probability determines the direct class prediction. For example, `[0.02784452, 0.97215548]` means there is a 2.78% probability of 'NotChurn' and a 97.22% probability of 'Churn', leading to a 'Churn' prediction.

### Data Analysis Key Findings
*   Direct class predictions (`y_pred_gbc`) output the most probable class label ('Churn' or 'NotChurn') for each customer.
*   Class probabilities (`y_pred_proba`) provide a probability distribution across classes for each prediction, with the first column indicating the probability of 'NotChurn' and the second column indicating the probability of 'Churn'.
*   The sum of probabilities for each instance across all classes in `y_pred_proba` is approximately 1.0.
*   The direct class prediction (e.g., 'Churn') is typically derived from the class with the highest probability in the `y_pred_proba` output (e.g., 97.22% probability for 'Churn' over 2.78% for 'NotChurn').

### Insights or Next Steps
*   Class probabilities provide a measure of the model's confidence in its predictions, which is more granular than just the direct class label.
*   These probabilities can be used to set custom classification thresholds, which is particularly useful in situations where the costs associated with false positives and false negatives are different.
